```python
import pandas as pd
import numpy as np

# Create your clean and correct dataset
data = {
    'Student_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Rahul Kumar', 'Amit Sharma', 'Priya Singh', 'Neha Verma', 'Ravi Kumar', 'Anjali Gupta', 'Vikas Singh', 'Pooja Sharma', 'Karan Mehta', 'Sneha Kumari'],
    'Gender': ['Male', 'Male', 'Female', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Department': ['CSE', 'CSE', 'ECE', 'CSE', 'ME', 'CSE', 'ECE', 'CSE', 'ME', 'ECE'],
    'Age': [20, 21, np.nan, 22, 23, 20, np.nan, 21, 22, 23],
    'Attendance': [85, 72, 90, np.nan, 75, 88, 92, np.nan, 76, 64],
    'Marks': [85, 72, np.nan, 91, 68, 88, 55, np.nan, 76, 64],
    'Email': ['rahul@gmail.com', 'amit@gmail.com', 'priya@gmail.com', 'neha@gmail.com', 'None', 'anjali@gmail.com', 'vikas@gmail.com', 'pooja@gmail.com', 'karan@gmail.com', 'sneha@gmail.com']
}

# Convert dictionary to Pandas DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
display(df)
```

```python
print('Shape:',df.shape)
print('Columns:',df.columns.tolist)
print('Data Types:',df.dtypes)
display(df.head())
display(df.tail())
```

```python
# Detect Missing values
print(df.isnull().sum())
```

```python
# Handle the missing values
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].median())
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['Email'] = df['Email'].fillna('Not Provided')

print(df.isnull().sum())
```

```python
# Data Transformation
df['Result'] = df['Marks'].apply(lambda x: 'Pass' if x >= 75 else 'Fail')

def assign_grade(marks):
    if marks >= 80:
        return 'A'
    elif marks >= 60:
        return 'B'
    elif marks >= 40:
        return 'C'
    elif marks >= 35:
        return 'D'
    else:
        return 'F'

df['Grade'] = df['Marks'].apply(assign_grade)
display(df)
```
